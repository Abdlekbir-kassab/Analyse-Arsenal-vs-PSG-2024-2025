```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample data (editable with real stats)
data = {
    'Statistic': [
        'Possession (%)',
        'Passes',
        'Passing Accuracy (%)',
        'Shots',
        'Shots on Target',
        'Tackles',
        'Interceptions'
    ],
    'Arsenal': [48.9, 520, 88.5, 13, 5, 45, 15],
    'Paris SG': [51.1, 350, 85.2, 8, 3, 38, 18]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Plot configuration
plt.figure(figsize=(10, 6))
indices = range(len(df))

# Bar settings
bar_width = 0.4
plt.barh([i + bar_width for i in indices], df['Arsenal'], height=bar_width, label='Arsenal', color='crimson')
plt.barh(indices, df['Paris SG'], height=bar_width, label='Paris SG', color='navy')

# Y-axis labels
plt.yticks([i + bar_width / 2 for i in indices], df['Statistic'])
plt.xlabel('Value')
plt.title('Team-Level Match Analysis: Arsenal vs Paris SG')
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Show plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Example xG values (replace with actual match data)
teams = ['Arsenal', 'Paris SG']
xg_values = [1.8, 1.2]  # Example: Arsenal had better chances

# Plot
plt.figure(figsize=(6, 4))
bars = plt.bar(teams, xg_values, color=['crimson', 'navy'])

# Add value labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.05, f'{height:.2f}', 
             ha='center', va='bottom', fontsize=12)

# Customize plot
plt.title('xG (Expected Goals) Comparison')
plt.ylabel('xG Value')
plt.ylim(0, max(xg_values) + 0.5)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Show
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Example xG event timeline (minute, cumulative xG) – you can replace these with actual match data
arsenal_minutes = [12, 29, 35, 38, 58]
arsenal_xg =     [0.05, 0.12, 0.40, 0.80, 1.20]

psg_minutes = [10, 15, 70, 80]
psg_xg =     [0.03, 0.08, 0.30, 0.95]

# Create step plot
plt.figure(figsize=(10, 5))

plt.step(arsenal_minutes, arsenal_xg, where='post', label='Arsenal', color='crimson')
plt.step(psg_minutes, psg_xg, where='post', label='Paris SG', color='navy')

# Half-time line
plt.axvline(x=45, color='gray', linestyle='--', alpha=0.5)

# Labels and title
plt.xlabel('Minute')
plt.ylabel('Cumulative xG')
plt.title('xG Timeline – Arsenal vs Paris SG')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

# Show plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dummy data for team performance stats
data = {
    'Statistic': [
        'Possession (%)',
        'Total Passes',
        'Pass Accuracy (%)',
        'Shots',
        'Shots on Target',
        'Fouls Committed',
        'Corners',
        'Offsides',
        'Tackles',
        'Interceptions'
    ],
    'Arsenal': [47, 480, 85, 14, 5, 10, 6, 2, 18, 11],
    'Paris SG': [53, 510, 88, 11, 4, 8, 4, 1, 15, 14]
}

# Create DataFrame
df = pd.DataFrame(data)

# Plot
plt.figure(figsize=(12, 7))
index = range(len(df))
bar_width = 0.4

# Horizontal bars
plt.barh([i + bar_width for i in index], df['Arsenal'], height=bar_width, label='Arsenal', color='red')
plt.barh(index, df['Paris SG'], height=bar_width, label='Paris SG', color='blue')

# Add labels and formatting
plt.yticks([i + bar_width / 2 for i in index], df['Statistic'])
plt.xlabel('Value')
plt.title('Team Statistics Comparison: Arsenal vs Paris SG')
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
